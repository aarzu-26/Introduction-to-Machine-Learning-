```python
import pandas as pd
# Sample dictionary to create DataFrame
data = {
    'name': ['Bob', 'Alice', 'Charlie', 'Eve', 'David'],
    'age': [25, 30, 35, 28, 22],
    'city': ['New York', 'Los Angeles', 'Chicago', 'Houston', 'Phoenix'],
    'salary': [70000, 80000, 120000, 65000, 90000]
}
```

```python
# ● Create a DataFrame from a dictionary and display it
df = pd.DataFrame(data)
print("Original DataFrame:")
print(df)
```

```python
# ● Sort the DataFrame by 'name' in ascending order
df_sorted = df.sort_values(by='name', ascending=True)
print("DataFrame sorted by 'name' in ascending order:")
print(df_sorted)

# Reset index after sorting (optional, for clean display) optional
df_sorted = df_sorted.reset_index(drop=True)
print("Sorted DataFrame with reset index:")
print(df_sorted)
```

```python
# ● Delete one specific column (e.g., 'city') from the DataFrame
df_dropped = df_sorted.drop('city', axis=1)
# Alternative ways:
# df_sorted.drop(columns='city', inplace=True)
# del df_sorted['city']
# df_sorted.pop('city')
print("DataFrame after deleting 'city' column:")
print(df_dropped)
```

```python
# ● Write the DataFrame to a CSV file using tab separator
df_dropped.to_csv('output_data.csv', index=False)
# Note: File will be saved as 'output_data.tsv' (common extension for tab-separated values)
print("DataFrame has been successfully written to 'output_data.tsv' using tab separator.")
```
