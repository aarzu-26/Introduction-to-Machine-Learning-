```python
# Practical No. 15
# Title: Simulate Green Hydrogen vs Ammonia Fuel Requirement for Cargo Ships
# Required Libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
# Step 1: Define Constants
HYDROGEN_ENERGY_DENSITY = 120  # MJ/kg
AMMONIA_ENERGY_DENSITY = 18.6  # MJ/kg
SHIP_ENERGY_REQUIREMENT = 1_000_000  # MJ
# Step 2: Calculate Fuel Requirement (Without Efficiency)
hydrogen_required = SHIP_ENERGY_REQUIREMENT / HYDROGEN_ENERGY_DENSITY
ammonia_required = SHIP_ENERGY_REQUIREMENT / AMMONIA_ENERGY_DENSITY
# Step 3: Create DataFrame
fuel_data = pd.DataFrame({
'Fuel Type': ['Green Hydrogen', 'Ammonia'],
'Energy Density (MJ/kg)': [HYDROGEN_ENERGY_DENSITY, AMMONIA_ENERGY_DENSITY],
'Fuel Required (kg)': [hydrogen_required, ammonia_required]
})
print("\nFuel Requirement Comparison:\n")
print(fuel_data)
# Step 4: Visualization
plt.figure()
plt.bar(fuel_data['Fuel Type'], fuel_data['Fuel Required (kg)'])
plt.xlabel('Fuel Type')
plt.ylabel('Fuel Required (kg)')
plt.title('Fuel Requirement Comparison for Cargo Ship')
plt.show()
# Step 5: Considering Efficiency
HYDROGEN_EFFICIENCY = 0.55
AMMONIA_EFFICIENCY = 0.40
hydrogen_actual = SHIP_ENERGY_REQUIREMENT / (HYDROGEN_ENERGY_DENSITY * HYDROGEN_EFFICIENCY)
ammonia_actual = SHIP_ENERGY_REQUIREMENT / (AMMONIA_ENERGY_DENSITY * AMMONIA_EFFICIENCY)
print("\nConsidering Engine Efficiency:\n")
print(f"Hydrogen Required (kg): {hydrogen_actual:.2f}")
print(f"Ammonia Required (kg): {ammonia_actual:.2f}")
```
