```python
#practical-4 
#simple 2-D array for demonstration
import numpy as np
arr_2d = np.array([[10,20,30,40],
                   [50,60,70,80],
                   [15,25,35,45],
                   [90,85,75,65]])
print ("Original 2D Array:")
print(arr_2d)
#1) to find the maximum and minimum value of given flattened array
flattened = arr_2d.flatten()   #converting the 2-d array into single dimensgion array
max_val = np.max(flattened)
min_val = np.min(flattened)

print("flattened array:")
print(flattened)    
print(f"maximum value in flattened array: {max_val}")
print(f"minimum value in flattened array: {min_val}")
```

```python
#2) to compute the mean,standard deviation and variance of a given array along the second axis
import numpy as np
mean_axis1 = np.mean(arr_2d , axis=1)
std_axis1 = np.std(arr_2d , axis=1)
var_axis1 = np.var(arr_2d , axis=1)

print("statistic along the second axis (row-wise):")
print("mean of each row:" , mean_axis1)
print("standard deviation of each row:", std_axis1)
print("variance of each row:" , var_axis1)
```
