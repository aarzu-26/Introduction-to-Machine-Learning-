```python
#1 : to convert a list of numeric values into one-dimensional numpy array
import numpy as np
a=[10,20,30,40]
array=np.array(a)
print(array)
```

```python
#2 : to create a 3*3 matrix with values ranging from 2 to 10
import numpy as np
values=np.arange(2,11)
print(values)
matrix=values.reshape(3,3)
print(matrix)
```

```python
#4: to split an array of 14 elements into 3 arrays, each with 2, 4, and 8 elements in the original order
#split 14 elements array into 3 arrays (2,4,8 arrays)
import numpy as np 
array=np.arange(14)
print("original array:",array)
split_arrays=np.split(array,[2,6])
print("splitted arrays:")
print(split_arrays)
for arr in split_arrays:
	print(arr)
```

```python
#4 : to stack arrays horizontally (column-wise)
import numpy as np
arr_a = np.array([[1],
                  [2],
                  [3]])
arr_b = np.array([[4],
                  [5],
                  [6]])
stacked_h = np.hstack((arr_a, arr_b))
print("Horizontal stack (column-wise) example 1:")
print(stacked_h)


import numpy as np
arr_c = np.array([[10, 20],
                  [30, 40]])
arr_d = np.array([[50, 60],
                  [70, 80]])
stacked_2d = np.hstack((arr_c, arr_d))
print("Horizontal stack example 2:")
print(stacked_2d)
```
