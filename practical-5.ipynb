```python
import numpy as np
import pandas as pd
# Sample NumPy array
np_array = np.array([45, 23, 67, 12, 89, 34, 56])
print("Original NumPy array:")
print(np_array)
```

```python
# ● Convert a NumPy array to a Pandas Series
series_from_array = pd.Series(np_array)
# Optionally provide custom index
series_from_array = pd.Series(np_array)
print("Pandas Series created from NumPy array:")
print(series_from_array)
```

```python
#● to create the mean and standard deviation of the data of a given Series
mean_value = series_from_array.mean()
std_value = series_from_array.std()
print("Statistical measures:")
print(f"Mean of the Series: {mean_value}")
print(f"Standard deviation of the Series: {std_value}")
```

```python
# ●to sort a given Series 
# Create another Series with unsorted data for sorting demonstration
unsorted_series = pd.Series([5, 2, 8, 1, 9, 3])
print("Unsorted Series:")
print(unsorted_series)

# ● Sort a given Series by values (ascending)
sorted_values_asc = unsorted_series.sort_values()
print("Series sorted by values (ascending):")
print(sorted_values_asc)
# Sort by values (descending)
sorted_values_desc = unsorted_series.sort_values(ascending=False)
print("Series sorted by values (descending):")
print(sorted_values_desc)
```
