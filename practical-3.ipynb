```python
#1) to add, subtract, multiply, divide arguments element wise
#sample arrays for demonstration
import numpy as np
arr1=np.array([10,20,30,40,50])
arr2= np.array([2,3,4,5,6])
print("array 1: ", arr1)
print("array 2:", arr2)
print()  #for space in next line 

#addition:
add_result = np.add(arr1, arr2)
print("element-wise addition:")
print(add_result)
print()  #for space in next line 

#subtraction:
sub_result = np.subtract(arr1,arr2)
print("element-wise subtraction:")
print(sub_result)
print()  #for space in next line 

#multiplication:
mul_result = np.multiply(arr1,arr2)
print("element-wise multiplication:")
print(mul_result)
print()   #for space in next line 

#division 
div_result = np.divide(arr1, arr2)
print("element-wise division:")
print(div_result)
print()

```

```python
#2) to round elements of the array to the nearest integer
import numpy as np
float_arr = np.array([1.2 , 2.7 , 3.5 , 4.1 , 5.9 , 6.3])
rounded = np.round(float_arr)
print("Rounded floating-point array:")
print(float_arr)
print("Rounded to nearest integer :")
print(rounded)
```

```python
#3) to calculate the difference between neighboring elements ,element-wise of given array
#difference between two arrays
import numpy as np
diff_arr = np.array([5,12,20,35,48,70])
differences = np.diff(diff_arr)
print("original array:")
print(diff_arr)
print("differences between neighboring elements:")
print(differences)
print()     #for space in next line

# difference between two matrix array 
matrix = np.array([[1,4,9],
                  [2,8,15],
                  [3,10,20]])
print("2D Matrix:")
print(matrix)
print("differences along axis=1 (row-wise):")
print(np.diff(matrix, axis=1))
print("Differences along axis=0 (column-wise):")
print(np.diff(matrix , axis=0))

```
