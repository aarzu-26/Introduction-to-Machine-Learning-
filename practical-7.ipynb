```python
import pandas as pd
import numpy as np

df=pd.read_csv('sample_data.csv')
print("Original DataFrame:")
print(df)
```

```python
#1) to find and drop the missing values from the given dataset 

# ● Find the missing values
print("Finding missing values (True indicates NaN):")
print(df.isnull())
print("Count of missing values in each column:")
print(df.isnull().sum())
print()
 
# ● Drop rows with any missing values (NaNs)
df_dropped_na = df.dropna()  # or df.dropna(how='any')
print("DataFrame after dropping rows with any NaNs:")
print(df_dropped_na)
```

```python
# Reset index after dropping rows (optional not given the manual)
df_dropped_na = df_dropped_na.reset_index(drop=True)
print("DataFrame after resetting index:")
print(df_dropped_na)
```

```python
# to remove the duplicate value from the given data set 
df_no_duplicates = df_dropped_na.drop_duplicates()
print("DataFrame after removing duplicate rows:")
print(df_no_duplicates)
```

```python
#drop rows with any NaNs
df_cleaned = df_no_duplicates.reset_index(drop=True)
print("Final cleaned DataFrame (duplicates removed + rows with any NaNs dropped):")
print(df_cleaned)
```
